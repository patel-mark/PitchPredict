```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/patel-mark/PitchPredict/refs/heads/main/football_stats_big5_expanded.csv')
df.head()
```

```python
df.columns
```

```python
df['yellow_cards_per_90']=(df['cards_yellow']/df['games']).round(2)
df['red_cards_per_90']=(df['cards_red']/df['games']).round(2)
df['progressive_carries_per_90']=(df['progressive_carries']/df['games']).round(2)
df['progressive_passes_per_90']=(df['progressive_passes']/df['games']).round(2)
df=df.drop(['cards_yellow','cards_red','progressive_carries','progressive_passes','possession','games','minutes','ranker','comp','players_used','avg_age','goals','games_starts','assists','minutes_90s',
       'goals_assists','goals_pens','pens_made','pens_att','xg','npxg','xg_assist','npxg_xg_assist','goals_assists_per90','goals_assists_pens_per90','xg_xg_assist_per90','npxg_xg_assist_per90'],axis=1)
```

```python
df = df.rename(columns={'goals_pens_per90':'non_pen_goals_per90',})
```
